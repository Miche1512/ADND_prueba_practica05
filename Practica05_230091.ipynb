## Ciencia de Datos para el Bien: DonorsChoose

DonorsChoose es una organización sin fines de lucro con sede en Estados Unidos que permite a las personas donar directamente a proyectos de aulas de escuelas públicas. Hasta enero de 2018, más de un millón de proyectos habían sido financiados a través de DonorsChoose.org. Además, en el 77 % de las escuelas públicas de Estados Unidos, al menos un proyecto ha sido solicitado mediante esta plataforma.

DonorsChoose ha compartido un conjunto de datos completo que incluye todos los proyectos realizados en el pasado, los recursos relacionados y los metadatos de docentes, escuelas, donantes y donaciones. Estos datos pueden utilizarse para descubrir información valiosa y desarrollar soluciones que permitan a DonorsChoose.org crear campañas de correo electrónico personalizadas, recomendando solicitudes específicas de aulas a donantes que ya han contribuido anteriormente.

En este cuaderno (notebook), se realizó un análisis y exploración exhaustivos del conjunto de datos compartido. Además de proporcionar información relevante sobre los datos de DonorsChoose, este notebook también funciona como un tutorial para aprender a realizar análisis de datos, análisis exploratorio de datos (EDA) y visualizaciones de datos.
